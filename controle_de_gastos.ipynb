```python
def ler_valor(texto):
    while True:
        try:
            valor = float(input(texto).replace(",", "."))
            if valor > 0:
                return valor
            print("Digite um valor maior que zero.")
        except ValueError:
            print("Valor inválido. Digite apenas números.")


print("===== CONTROLE DE GASTOS =====")

orcamento = ler_valor("Digite seu orçamento do mês: R$ ")

limite = orcamento * 0.80    # máximo que pode gastar (80%)
reserva = orcamento * 0.20   # mínimo que precisa sobrar (20%)

print(f"\nOrçamento registrado: R$ {orcamento:.2f}")
print(f"Você pode gastar no máximo R$ {limite:.2f}")
print(f"Reserva obrigatória (20%): R$ {reserva:.2f}")

total = 0

while True:
    print("\n===== MENU =====")
    print("1 - Registrar gasto")
    print("2 - Ver quanto já gastou")
    print("3 - Verificar orçamento")
    print("4 - Sair")

    opcao = input("Digite uma opção: ")

    if opcao == "1":
        gasto = ler_valor("Digite o valor do gasto: R$ ")

        if total + gasto > limite:
            print("Gasto NÃO registrado! Você ficaria com menos de 20% do orçamento.")
            print(f"Ainda pode gastar até R$ {limite - total:.2f}")
        else:
            total = total + gasto
            print("Gasto registrado com sucesso!")

    elif opcao == "2":
        print(f"Você já gastou R$ {total:.2f}")

    elif opcao == "3":
        print(f"Orçamento: R$ {orcamento:.2f}")
        print(f"Já gastou: R$ {total:.2f}")
        print(f"Ainda pode gastar: R$ {limite - total:.2f}")
        print(f"Reserva mantida: R$ {orcamento - total:.2f}")

    elif opcao == "4":
        print("Programa encerrado!")
        break

    else:
        print("Opção inválida.")
```
